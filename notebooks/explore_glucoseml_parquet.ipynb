## Summary Statistics Across All Datasets


```python
# Create summary dataframe
summary_data = []

for file in parquet_files:
    df = pl.read_parquet(file)
    summary_data.append({
        "dataset": file.stem,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "file_size_mb": file.stat().st_size / (1024 * 1024)
    })

summary_df = pl.DataFrame(summary_data)
display(summary_df)
```

## Check for Common Columns Across Datasets


```python
# Load all datasets and get their column names
all_columns = {}
for file in parquet_files:
    df = pl.read_parquet(file)
    all_columns[file.stem] = set(df.columns)

# Find common columns
common_cols = set.intersection(*all_columns.values())

# Create summary of column info
column_summary = {
    "common_columns": sorted(common_cols),
    "datasets": {}
}

for dataset_name, cols in all_columns.items():
    unique_cols = cols - common_cols
    column_summary["datasets"][dataset_name] = {
        "all_columns": sorted(cols),
        "unique_columns": sorted(unique_cols) if unique_cols else []
    }

display(column_summary)
```
